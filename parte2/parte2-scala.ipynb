## Ejercicio 1 — Variables, tipos e inferencia
### Planteamiento
En este ejercicio modelamos la información de un jugador en una partida de cartas a través de dos aproximaciones:
1. **Versión A (Tipado explícito):** Declaramos cada identificador indicando de forma explícita su tipo de dato (`String`, `Int`, `Double`, `Boolean`).
2. **Versión B (Inferencia de tipos):** Declaramos las variables omitiendo la anotación de tipo para que el compilador de Scala deduzca automáticamente el tipo en función del literal asignado.

```scala
// ── Versión A: tipos explícitos ──────────────────────────────

// Versión A: Tipado explícito
val nombreJugadorA: String = "Juan"
val partidasJugadasA: Int = 12
val puntuacionActualA: Int = 1500
val porcentajeVictoriasA: Double = 75.5
val estaActivoA: Boolean = true

println("=== Versión A (Tipado explícito) ===")
println(s"Jugador: $nombreJugadorA")
println(s"Partidas jugadas: $partidasJugadasA")
println(s"Puntuación actual: $puntuacionActualA")
println(s"Porcentaje de victorias: $porcentajeVictoriasA%")
println(s"¿Sigue activo?: $estaActivoA")
```

```scala
// ── Versión B: inferencia de tipos ───────────────────────────

val nombreJugadorB = "Juan"
val partidasJugadasB = 12
val puntuacionActualB = 1500
val porcentajeVictoriasB = 75.5
val estaActivoB = true

println("=== Versión B (Inferencia de tipos) ===")
println(s"Jugador: $nombreJugadorB")
println(s"Partidas jugadas: $partidasJugadasB")
println(s"Puntuación actual: $puntuacionActualB")
println(s"Porcentaje de victorias: $porcentajeVictoriasB%")
println(s"¿Sigue activo?: $estaActivoB")
```

```scala
## Ejercicio 2 — val, var y reasignación

### Planteamiento
Analizamos el principio de inmutabilidad en Scala contrastando variables mutables (`var`) 
frente a inmutables (`val`). Simularemos la actualización sucesiva de una puntuación y forzaremos 
un error de compilación intencionado al intentar reasignar una referencia inmutable.
```

```scala
val jugador = "Marta"
var puntuacion = 10

println(s"Jugador inicial: $jugador")
println(s"Puntuación inicial: $puntuacion")

// Incremento en 5 puntos
puntuacion = puntuacion + 5
println(s"Puntuación tras primer incremento (+5): $puntuacion")

// Incremento en 3 puntos
puntuacion = puntuacion + 3
println(s"Puntuación final (+3): $puntuacion")
```
